```python
negative_pairs = []

for _, s1_row in source1.head(10).iterrows():
    s1_id = s1_row["entity_id"]

    true_matches = ground_truth[ground_truth["source1_entity_id"] == s1_id]["matched_entity_ids"].iloc[0]

    true_matches = (
        []
        if pd.isna(true_matches)
        else true_matches.split(",")
    )

    for _, candidate_row in source23.iterrows():
        candidate_id = candidate_row["entity_id"]

        if candidate_id not in true_matches:

            negative_pairs.append(
                {
                    "source1_entity_id": s1_id,
                    "matched_entity_id": candidate_id,
                    "business_name_s1": s1_row["business_name"],
                    "business_name_matched": candidate_row["business_name"],
                    "business_address_s1": s1_row["business_address"],
                    "business_address_matched": candidate_row["business_address"],
                    "country_s1": s1_row["country"],
                    "country_matched": candidate_row["country"]
                }
            )

            break

    if len(negative_pairs) >= 10:
        break
```

```python
negative_pairs = pd.DataFrame(negative_pairs)
negative_pairs.head()
```

```python
negative_features = []

for _, row in pd.DataFrame(negative_pairs).iterrows():
    name_1 = row["business_name_s1"]
    name_2 = row["business_name_matched"]

    address_1 = row["business_address_s1"]
    address_2 = row["business_address_matched"]

    country_1 = row["country_s1"]
    country_2 = row["country_matched"]

    name_similarity = ratio(name_1,name_2)

    if pd.isna(address_1) or pd.isna(address_2):
        address_similarity = None

    else:
        address_similarity = ratio(address_1,address_2)


    country_same = int(country_1 == country_2)
    address_1_missing = int(pd.isna(address_1))
    address_2_missing = int(pd.isna(address_2))

    negative_features.append(
        {
            "source1_entity_id": row["source1_entity_id"],
            "matched_entity_id": row["matched_entity_id"],
            "label": 0,
            "name_similarity": name_similarity,
            "address_similarity": address_similarity,
            "country_same": country_same,
            "address_1_missing": address_1_missing,
            "address_2_missing": address_2_missing
        }
    )

negative_features = pd.DataFrame(
    negative_features
)

negative_features
```
