```python
import time

from business_entity_resolution.components.data_ingestion import DataIngestion
from business_entity_resolution.components.data_transformation import DataTransformation
from business_entity_resolution.components.candidate_generation import CandidateGeneration

# 1. Load full training data
ingestion = DataIngestion()
source1, source2, source3, ground_truth = ingestion.initiate_data_ingestion()

print("Loaded:")
print("S1:", source1.shape)
print("S2:", source2.shape)
print("S3:", source3.shape)

# 2. Transform full data
transformer = DataTransformation()

start = time.time()

source1 = transformer.transform_source(source1)
source2 = transformer.transform_source(source2)
source3 = transformer.transform_source(source3)

print(f"\nTransformation time: {time.time() - start:.2f} seconds")

# 3. 1,000 S1 benchmark against FULL S2/S3
benchmark_s1 = source1.sample(
    n=1000,
    random_state=42
).copy()

print("\nBenchmark:")
print("S1:", benchmark_s1.shape)
print("S2:", source2.shape)
print("S3:", source3.shape)

# 4. Candidate generation
candidate_generator = CandidateGeneration()

start = time.time()

candidates_s2, candidates_s3 = (
    candidate_generator.generate_candidates(
        benchmark_s1,
        source2,
        source3
    )
)

elapsed = time.time() - start

print("\n====================================")
print("BLOCKING BENCHMARK")
print("====================================")
print("S2 candidates:", len(candidates_s2))
print("S3 candidates:", len(candidates_s3))
print(f"Time: {elapsed:.2f} seconds")
print("====================================")
```
